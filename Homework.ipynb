```python
# 📘 COMMON DATASET (10 Students)
# =========================================
students_data = [
    (1, "Sophea", 89),
    (2, "Ratha", 76),
    (3, "Chenda", 92),
    (4, "Dara", 85),
    (5, "Sokun", 78),
    (6, "Malis", 91),
    (7, "Piseth", 67),
    (8, "information", 88),
    (9, "Ice tea", 73),
    (10, "jee hong", 95)
]
```

```python
# 🎓 SINGLY LINKED LIST
# =========================================

class SinglyNode:
    def __init__(self, id, name, score):
        self.id = id
        self.name = name
        self.score = score
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def add(self, id, name, score):
        new_node = SinglyNode(id, name, score)
        if not self.head:
            self.head = new_node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node

    def display(self):
        print("\n🎓 Singly Linked List:")
        current = self.head
        while current:
            print(f"ID:{current.id}, Name:{current.name}, Score:{current.score}")
            current = current.next

# Build and display
singly_list = SinglyLinkedList()
for s in students_data:
    singly_list.add(*s)
singly_list.display()
```

```python
# 🔁 DOUBLY LINKED LIST
# =========================================

class DoublyNode:
    def __init__(self, id, name, score):
        self.id = id
        self.name = name
        self.score = score
        self.prev = None
        self.next = None

class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def add(self, id, name, score):
        new_node = DoublyNode(id, name, score)
        if not self.head:
            self.head = self.tail = new_node
        else:
            self.tail.next = new_node
            new_node.prev = self.tail
            self.tail = new_node

    def display_forward(self):
        print("\n➡️ Doubly Linked List (Forward):")
        current = self.head
        while current:
            print(f"ID:{current.id}, Name:{current.name}, Score:{current.score}")
            current = current.next

    def display_backward(self):
        print("\n⬅️ Doubly Linked List (Backward):")
        current = self.tail
        while current:
            print(f"ID:{current.id}, Name:{current.name}, Score:{current.score}")
            current = current.prev

# Build and display
doubly_list = DoublyLinkedList()
for s in students_data:
    doubly_list.add(*s)
doubly_list.display_forward()
doubly_list.display_backward()
```

```python
# 🔄 CIRCULAR LINKED LIST
# =========================================

class CircularNode:
    def __init__(self, id, name, score):
        self.id = id
        self.name = name
        self.score = score
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.head = None

    def add(self, id, name, score):
        new_node = CircularNode(id, name, score)
        if not self.head:
            self.head = new_node
            new_node.next = new_node
        else:
            current = self.head
            while current.next != self.head:
                current = current.next
            current.next = new_node
            new_node.next = self.head

    def display(self, rounds=1):
        if not self.head:
            return
        print("\n🔁 Circular Linked List:")
        current = self.head
        count = 0
        while count < rounds:
            print(f"ID:{current.id}, Name:{current.name}, Score:{current.score}")
            current = current.next
            if current == self.head:
                count += 1

# Build and display
circular_list = CircularLinkedList()
for s in students_data:
    circular_list.add(*s)
circular_list.display(rounds=1)
```
